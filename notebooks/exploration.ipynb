# MindMove exploration

Start Jupyter from the project root after `python -m mindmove.data --config configs/default.yaml`.

Runs 4, 8, and 12 are imagined left/right hand movement. `T1` is left, `T2` is right, and `T0` (rest) is excluded. Other runs use different meanings for `T1` and `T2` and are not loaded here.


```python
from pathlib import Path
import sys

ROOT = Path.cwd()
if not (ROOT / "app.py").exists() and (ROOT.parent / "app.py").exists():
    ROOT = ROOT.parent
sys.path.insert(0, str(ROOT / "src"))

from mindmove.config import load_config
from mindmove.data import load_cached_dataset, split_table

cfg = load_config(ROOT / "configs" / "default.yaml")
table, manifest = load_cached_dataset(cfg)
train, val, test = split_table(table, cfg)
print("train", train.distribution())
print("validation", val.distribution())
print("test", test.distribution())
print("excluded", len(manifest.get("exclusions", [])))
print("observed Hz", manifest.get("observed_sfreq_hz"), "resampled", manifest.get("resampled_any"))
```

```python
import numpy as np
from mindmove.preprocessing import SENSORIMOTOR_CHANNELS, display_name
from mindmove.visualization import plot_trial_waveforms

index = int(np.flatnonzero(test.labels == 0)[0])
names = list(test.ch_names)
chosen = [name for name in SENSORIMOTOR_CHANNELS if name in names]
picked = test.data[index, [names.index(name) for name in chosen]]
title = f"{test.trial_ids()[index]}  {display_name(int(test.labels[index]))}"
plot_trial_waveforms(test.times, picked, chosen, title)
```
